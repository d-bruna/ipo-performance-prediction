```python
import pandas as pd
import numpy as np

# Load data
df = pd.read_csv('data_with_1year_returns_FINAL.csv')
df.head()
```

```python
completed = df[df['status'] == 'success'].copy()

print(f"Total IPOs with data: {len(completed)}")
print(f"\nCurrent statistics:")
print(f"  Mean:   {completed['1_year_return'].mean():+.2f}%")
print(f"  Median: {completed['1_year_return'].median():+.2f}%")
print(f"  Max:    {completed['1_year_return'].max():+.2f}%")
```

```python
outliers = completed[completed['1_year_return'] > 1000]

print(f"\n" + "="*80)
print(f"OUTLIERS TO REMOVE (Return > 1000%)")
print("="*80)
print(f"\nFound {len(outliers)} outliers:\n")

if len(outliers) > 0:
    print(outliers[['Issuer Ticker', 'Issuer Name', 'Offer Price', 'price_1yr', '1_year_return']].to_string())

# Mark outliers in dataframe
df.loc[df['1_year_return'] > 1000, 'status'] = 'outlier_removed'

# Get new statistics
cleaned = df[df['status'] == 'success'].copy()
```

```python
if len(cleaned) > 0:
    positive = (cleaned['1_year_up'] == 1).sum()
    print(f"\nNew statistics:")
    print(f"  Mean:   {cleaned['1_year_return'].mean():+.2f}%")
    print(f"  Median: {cleaned['1_year_return'].median():+.2f}%")
    print(f"  Std:    {cleaned['1_year_return'].std():.2f}%")
    print(f"  Min:    {cleaned['1_year_return'].min():+.2f}%")
    print(f"  Max:    {cleaned['1_year_return'].max():+.2f}%")
    print(f"\n  Positive: {positive} ({positive/len(cleaned)*100:.1f}%)")
    print(f"  Negative: {len(cleaned)-positive} ({(len(cleaned)-positive)/len(cleaned)*100:.1f}%)")

# Save cleaned data
df.to_csv('data_with_1year_returns_CLEANED.csv', index=False)
```

```python
import pandas as pd

# Load the cleaned data
df = pd.read_csv('data_with_1year_returns_CLEANED.csv')

print("="*80)
print("CREATING FINAL CSV - ONLY IPOs WITH 1-YEAR DATA")
print("="*80 + "\n")

# Filter: Only keep rows with successful 1-year return data
df_final = df[df['status'] == 'success'].copy()

print(f"Original dataset: {len(df)} rows")
print(f"With 1-year data: {len(df_final)} rows")
print(f"Removed: {len(df) - len(df_final)} rows\n")

# Show what we're keeping
print("Columns in final dataset:")
print(df_final.columns.tolist())

# Check for any missing values in key columns
print(f"\nMissing values check:")
print(f"  1_year_return: {df_final['1_year_return'].isna().sum()}")
print(f"  1_year_up: {df_final['1_year_up'].isna().sum()}")
print(f"  price_1yr: {df_final['price_1yr'].isna().sum()}")

# Statistics
print(f"\n1-Year Return Statistics:")
print(f"  Mean:   {df_final['1_year_return'].mean():+.2f}%")
print(f"  Median: {df_final['1_year_return'].median():+.2f}%")
print(f"  Std:    {df_final['1_year_return'].std():.2f}%")
print(f"  Min:    {df_final['1_year_return'].min():+.2f}%")
print(f"  Max:    {df_final['1_year_return'].max():+.2f}%")

positive = (df_final['1_year_up'] == 1).sum()
print(f"\n  Positive: {positive} ({positive/len(df_final)*100:.1f}%)")
print(f"  Negative: {len(df_final)-positive} ({(len(df_final)-positive)/len(df_final)*100:.1f}%)")

# Save final CSV
output_file = 'ipo_data_with_1year_returns_FINAL.csv'
df_final.to_csv(output_file, index=False)

print(f"\n✓ Saved to: {output_file}")
print(f"\nThis file contains {len(df_final)} IPOs with complete 1-year return data")
print("Ready for ML models! 🚀")

# Show first few rows
print(f"\nFirst 5 rows preview:")
print(df_final[['Issuer Ticker', 'Issuer Name', 'Offer Price', 'price_1yr', '1_year_return', '1_year_up']].head().to_string())

print("\n✅ Done!")
```
